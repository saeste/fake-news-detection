```python
import numpy as np
import scipy.sparse

from sklearn.model_selection import GridSearchCV

from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

# 4.1 Učitavanje TF-IDF podataka

```python
X_train = scipy.sparse.load_npz("data/X_train.npz")
X_test = scipy.sparse.load_npz("data/X_test.npz")

y_train = np.load("data/y_train.npy")
y_test = np.load("data/y_test.npy")

print("Trening skup:", X_train.shape)
print("Test skup:", X_test.shape)
```

# 4.2 Naive Bayes

```python
print("\nNAIVE BAYES")

nb_param_grid = {
    "alpha": [0.01, 0.1, 0.5, 1.0, 2.0]
}

nb_grid = GridSearchCV(
    estimator=MultinomialNB(),
    param_grid=nb_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

nb_grid.fit(X_train, y_train)
```

# 4.3 Random Forest

```python
print("\nRANDOM FOREST")

rf_param_grid = {
    "n_estimators": [50],
    "max_depth": [10, 20]
}

rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),
    param_grid=rf_param_grid,
    cv=3,
    scoring="accuracy",
    n_jobs=-1
)
rf_grid.fit(X_train, y_train)
```

```python
import joblib
joblib.dump(nb_grid, 'data/nb_model.joblib')
joblib.dump(rf_grid, 'data/rf_model.joblib')
```
